```python
import pandas as pd


cols_texto = {'description': 'Sin información', 'host_location': 'Sin información', 'host_about': 'Sin información'}
df = df.fillna(value=cols_texto)

cols_reviews_cero = [
    'first_review', 'last_review', 'review_scores_rating', 'review_scores_accuracy',
    'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication',
    'review_scores_location', 'review_scores_value', 'reviews_per_month'
]
df[cols_reviews_cero] = df[cols_reviews_cero].fillna(0)


cols_mediana = [
    'bedrooms', 'bathrooms', 'beds', 'price', 'estimated_revenue_l365d',
    'price_quote_total_price', 'price_quote_price_per_night', 'price_quote_raw'
]
for col in cols_mediana:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        df[col] = df[col].fillna(df[col].median())


cols_promedio = [
    'hosts_time_as_user_years', 'hosts_time_as_user_months',
    'hosts_time_as_host_years', 'hosts_time_as_host_months'
]
for col in cols_promedio:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())


cols_propagacion = [
    'host_profile_id', 'host_profile_url', 'host_name', 'host_is_superhost',
    'host_picture_url', 'host_listings_count', 'host_has_profile_pic',
    'host_identity_verified', 'bathrooms_text', 'minimum_nights', 'maximum_nights',
    'minimum_minimum_nights', 'maximum_minimum_nights', 'minimum_maximum_nights',
    'maximum_maximum_nights', 'has_availability',
    'price_quote_checkin_date', 'price_quote_checkout_date'
]
cols_existentes_prop = [col for col in cols_propagacion if col in df.columns]
df[cols_existentes_prop] = df[cols_existentes_prop].ffill().bfill()


cols_100_nulas = [
    'neighborhood_overview', 'host_since', 'host_response_time', 'host_response_rate',
    'host_acceptance_rate', 'host_thumbnail_url', 'host_neighbourhood', 
    'host_total_listings_count', 'host_verifications', 'neighbourhood', 
    'neighbourhood_group_cleansed', 'calendar_updated', 'license', 'instant_bookable'
]
df = df.drop(columns=cols_100_nulas, errors='ignore')
# Verificar cuántos nulos quedan en todo el dataset (debe dar 0)
print("Total de valores nulos restantes:", df.isnull().sum().sum())
```
